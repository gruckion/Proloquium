```python
import jedi
import ast

def get_ast(file_path):
    with open(file_path, 'r') as file:
        source_code = file.read()

    script = jedi.Script(source_code)
    parsed_ast = ast.parse(source_code)

    return parsed_ast

# Example usage
file_path = '/Users/sigex/workdir/proloquium/src/app.py'
ast_tree = get_ast(file_path)

# Printing the AST
ast.dump(ast_tree)
```

```python
import jedi

file_path = '/Users/sigex/workdir/proloquium/src/app.py'

def analyze_file(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        first_line = file.readline()
        docstring = first_line.strip().lstrip('#').strip()

    definitions = jedi.Script(code=first_line, path=file_path).goto_definitions()
    definitions_names = [definition.name for definition in definitions]

    references = jedi.Script(code=first_line, path=file_path).get_references()
    references_names = [reference.name for reference in references]

    print(f"Definitions: {definitions_names}")
    print(f"References: {references_names}")
    print(f"Docstring: {docstring}" if docstring else "No docstring found.")

analyze_file(file_path)
```
